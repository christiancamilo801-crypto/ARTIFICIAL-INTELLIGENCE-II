Resumen PCA

PCA (Análisis de Componentes Principales) es una técnica que permite reducir la cantidad de variables de un conjunto de datos conservando la mayor cantidad posible de información.

El proceso básico es:

Organizar los datos. Estandarizar las variables para que tengan escalas comparables. Calcular la matriz de covarianza para conocer las relaciones entre variables. Obtener autovalores y autovectores, que determinan los componentes principales. Seleccionar los componentes que conserven la mayor parte de la información. Transformar los datos a las nuevas dimensiones.

Ejemplo: si tenemos 10 variables, PCA podría reducirlas a 2 o 3 componentes principales manteniendo gran parte de la información original.

Objetivo principal: reducir dimensiones, facilitar el análisis y visualizar datos más fácilmente.
